```python
# DataFrame 

df = pd.DataFrame(data.data, columns = data.feature_names)
df.head()
```

```python
df['Price'] = data.target
```

```python
df
```

```python
df1 = df.sample(frac = 0.20)
#to visualise the tree, selecting only 20 percent of the data, feel free to comment this line to train the model
```

```python
df1
```

```python
#Separate X and y
X = df1.iloc[:, :-1]
y = df1.iloc[:, -1]
```

```python
#train - test - split

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 1)
```

```python
X_train.shape, X_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
# Model Training 

from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test, y_pred))
```

```python
#hyperparameter tuning
params = {
    'criterion': ["squared_error", "absolute_error", "friedman_mse", 'poisson'],
    'splitter': ["best", "random"],
    'max_depth': [1, 2, 3, 4, 10],
    'max_features':['auto', 'sqrt', 'log2']
}
```

```python
# Hyperparameter Tuning with Cv

from sklearn.model_selection import GridSearchCV
reg = DecisionTreeRegressor()
model = GridSearchCV(reg, param_grid=params, cv=5,  scoring='neg_mean_squared_error', verbose=3)
model.fit(X_train, y_train)
```
