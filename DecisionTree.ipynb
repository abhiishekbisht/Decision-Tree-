```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

import warnings 
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris
data = load_iris()
```

```python
data.data
```

```python
data.target
```

```python
print(data.DESCR)
```

```python
df = pd.DataFrame(data.data, columns = data.feature_names)
df.head()
```

```python
df['target'] = data.target
```

```python
df
```

```python
# Split X and y

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

```

```python
# train - test split

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.30, random_state = 1)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(criterion = 'entropy')
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn import tree
plt.figure(figsize = (10, 8))
tree.plot_tree(model, filled = True)
# plt.show() # to hide entropy details 
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
#post pruning >> after making complete decision tree you have to cut it to avoid overfitting
#for small dataset only

#observation
#observe the decision tree plot, you will realise the last split is actually overfitting, diving the data 
#in such as way that each node is 100 percent pure leading to low no of data point, i.e-1 and 3
#so reduce the depth of dt
```

```python
from sklearn.tree import DecisionTreeClassifier
post_pruend_model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 2)
post_pruend_model.fit(X_train, y_train)
```

```python
from sklearn import tree
plt.figure(figsize = (10, 8))
tree.plot_tree(post_pruend_model, filled = True)
```

```python
y_pred_pruend = post_pruend_model.predict(X_test)
y_pred_pruend
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred_pruend)
```

```python
#the accuracy is same, earlier it was a overfitting model
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_pruend))
```

```python
# Pre-pruning Using HyperParameter Tuning - 
```

```python
#prepruning using hyperparameter tuning

parameter = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'splitter' : ['best', 'random'],
    'max_depth': [1, 2, 3, 4, 5, 6, 7, 10],
    'max_features':['sqrt', 'log2', 'auto']
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier()

model = GridSearchCV(clf, param_grid = parameter, cv = 5, scoring = 'accuracy', verbose = 3)
```

```python
model.fit(X_train, y_train)
```

```python
model.best_params_
```

```python
model.best_score_
```
